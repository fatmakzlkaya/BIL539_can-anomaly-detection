```python
import pandas as pd
import os
```

```python
repo_root = os.path.dirname(os.path.dirname(os.path.abspath("__file__")))  # repo root
normal_file = os.path.join(repo_root, 'data/raw/10_second_master_and_slave.csv')
broken_file = os.path.join(repo_root, 'data/raw/10_second_master_and_slave_broken.csv')
```

```python
df_normal = pd.read_csv(normal_file)
df_broken = pd.read_csv(broken_file)
```

```python
# Drop the unnamed column
df_normal = df_normal.loc[:, ~df_normal.columns.str.contains('^Unnamed')]
df_broken = df_broken.loc[:, ~df_broken.columns.str.contains('^Unnamed')]
```

```python
payload_cols = ['d0','d1','d2','d3','d4','d5','d6','d7']
```

```python
for col in payload_cols:
    df_normal[col] = df_normal[col].apply(lambda x: int(x,16))
    df_broken[col] = df_broken[col].apply(lambda x: int(x,16))
```

```python
df_normal['label'] = 0
df_broken['label'] = 1
```

```python
df = pd.concat([df_normal, df_broken], ignore_index=True)
```

```python
df = df.sort_values('timestamp').reset_index(drop=True)
```

```python
df['interarrival'] = df['timestamp'].diff().fillna(0)
```
